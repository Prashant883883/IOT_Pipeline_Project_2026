# RDI Recruiter Module - ML Integration Documentation

## Overview
This document provides comprehensive documentation of the Machine Learning features integrated into the RDI Recruiter Module. The ML system is designed to help recruiters make data-driven decisions through predictive analytics, intelligent matching, and recruitment forecasting.

### Key Features
1. **Advanced Skill Matching** - Semantic skill comparison using fuzzy matching
2. **Hiring Success Prediction** - ML model predicting probability of successful hire
3. **Recruitment Analytics** - Comprehensive KPI tracking and insights
4. **Population Forecasting** - Predict future hiring trends and pipeline state
5. **Candidate Clustering** - Segment candidates by performance tiers
6. **Anomaly Detection** - Identify unusual recruitment patterns
7. **Skill Demand Intelligence** - Track trending skills and market demand

## Architecture Overview

### System Design
The ML system is built as a backend API-first architecture with the following components:

```
Frontend (React/Next.js)
    ↓
ML Insights Page
    ↓
API Routes (/api/ml/*)
    ↓
ML Analytics Functions (lib/ml-analytics.ts)
    ↓
Database (Prisma)
```

### Data Flow
1. **Data Collection**: Applications, proposals, and student data stored in database
2. **Feature Engineering**: Raw data processed into meaningful features
3. **ML Processing**: Advanced algorithms applied for predictions and analysis
4. **API Response**: Results formatted and returned to frontend
5. **Visualization**: Results displayed in dashboards and insights pages

### Key Technologies
- **Language**: TypeScript/Node.js
- **Framework**: Next.js (API routes)
- **Database**: SQLite (with Prisma ORM)
- **ML Algorithms**: Custom implementations of:
  - Fuzzy string matching (Levenshtein distance)
  - K-means clustering
  - Linear trend forecasting
  - Anomaly detection algorithms
  - Weighted scoring models

## API Endpoints Reference

### 1. Advanced Skill Matching
**Endpoint**: `GET /api/ml/advanced-matching`

**Purpose**: Perform detailed skill matching between student and proposal with semantic similarity analysis.

**Query Parameters**:
```
- studentId: UUID of student
- proposalId: UUID of proposal
OR
- studentSkills: Comma-separated skills (e.g., "Python, JavaScript, React")
- proposalSkills: Comma-separated required skills
```

**Response Example**:
```json
{
  "success": true,
  "data": {
    "studentName": "John Doe",
    "proposalTitle": "Senior Frontend Developer",
    "matchingAnalysis": {
      "overallScore": 85,
      "matchPercentage": 85,
      "exactMatchCount": 5,
      "partialMatchCount": 2,
      "missingSkillCount": 1,
      "totalRequiredSkills": 8
    },
    "detailedBreakdown": {
      "exactMatches": [
        {"skill": "React", "type": "exact", "confidence": 100}
      ],
      "partialMatches": [
        {"requiredSkill": "TypeScript", "candidateSkill": "JavaScript", "similarity": 75}
      ],
      "missingSkills": [
        {"skill": "GraphQL", "type": "missing"}
      ]
    },
    "recommendation": {
      "level": "good",
      "message": "👍 Good match - recommended for interview"
    }
  }
}
```

### 2. Hiring Success Prediction
**Endpoint**: `GET /api/ml/hiring-prediction?applicationId={id}`

**Purpose**: Predict the probability that a candidate will be successfully hired.

**Response Example**:
```json
{
  "success": true,
  "data": {
    "applicationId": "app-123",
    "studentName": "Jane Smith",
    "proposalTitle": "ML Engineer",
    "successProbability": 78,
    "confidenceLevel": "high",
    "recommendation": "⭐ Moderately likely - schedule interview",
    "keyFactors": [
      {"factor": "Skill Match", "impact": 25, "direction": "positive"},
      {"factor": "Application Momentum", "impact": 15, "direction": "positive"},
      {"factor": "Recruiter Engagement", "impact": 18, "direction": "positive"},
      {"factor": "Historical Success Rate", "impact": 12, "direction": "positive"},
      {"factor": "Pipeline Stage", "impact": 8, "direction": "positive"}
    ]
  }
}
```

### 3. Recommendations

**Query Parameters**:

## Core ML Functions

### Advanced Skill Matching Functions

#### 1. advancedSkillMatch()
Performs semantic skill matching with fuzzy string matching.

**Algorithm**:
1. Normalize all skills (lowercase, trim)
2. Find exact matches
3. For non-exact matches, calculate Levenshtein similarity
4. Apply 50% similarity threshold
5. Calculate overall score: (exact + 0.7 × partial) / required

**Key Metrics**:
- Exact matches: Skills that match exactly
- Partial matches: Similar skills above threshold
- Missing skills: Required skills not found
- Match percentage: Percentage of matched requirements

#### 2. calculateSkillSimilarity()
Calculates similarity between two skills using multiple methods:
- Exact match (score = 1.0)
- Substring match (score = 0.85)
- Levenshtein distance (score based on edit distance)

#### 3. Levenshtein Distance Algorithm
Dynamic programming algorithm that calculates minimum edits needed to transform one string to another.

**Time Complexity**: O(m × n) where m, n are string lengths
**Use Case**: Matching "JavaScript" with "JS" or typos

---

### Predictive Models

#### 4. predictHiringSuccess()
Predicts hiring success probability using weighted factors:

**Factors** (weights):
1. Skill Match: 30%
2. Application Momentum: 20%
   - Recent applications weighted higher
3. Recruiter Engagement: 20%
   - Number of recruiter notes
4. Historical Success Rate: 15%
   - Past hiring success of candidate
5. Pipeline Stage: 15%
   - Current stage in hiring funnel

**Output**: 
- Success probability (0-100%)
- Confidence level (low/medium/high)
- Key contributing factors

#### 5. forecastApplicationTrend()
Linear trend forecasting for application volume.

**Method**:
1. Group applications by month
2. Calculate moving average
3. Compute trend line
4. Extrapolate for future months
5. Add slight seasonal adjustment

**Forecast Period**: 6-24 months ahead

#### 6. forecastPipelineState()
Predicts future hiring, rejections, and in-progress applications.

**Features**:
- Seasonality adjustment (sine wave)
- Confidence intervals based on historical data
- Separate forecasts for each pipeline stage

---

### Clustering & Segmentation

#### 7. clusterCandidates()
K-means style clustering of candidates by performance.

**Clusters**:
1. **High Performers** (75-100%): Ready to hire
2. **Strong Candidates** (50-75%): Good fit
3. **Potential Candidates** (25-50%): Need development
4. **Development Needed** (0-25%): Significant gaps

#### 8. findSimilarCandidates()
Finds similar candidates based on skill overlap.

**Similarity Threshold**: 60% by default (configurable)

**Returns**:
- Candidate ID
- Similarity score (0-100%)
- Common skills
- Complementary skills

---

### Anomaly Detection

#### 9. detectAnomalies()
Identifies unusual recruitment patterns.

**Detects**:
1. **Application Drop**: Recent submissions significantly below average
2. **High Rejection Rate**: >60% rejection rate
3. **Stalled Applications**: Applications inactive >30 days

**Severity Levels**: low, medium, high

## Usage Examples

### Example 1: Advanced Skill Matching

**Scenario**: Match a candidate's skills against a job proposal.

**Request**:
```
GET /api/ml/advanced-matching?studentId=abc123&proposalId=def456
```

**Frontend Usage**:
```javascript
// Check skill match for a specific application
const response = await fetch(
  `/api/ml/advanced-matching?studentId=${studentId}&proposalId=${proposalId}`
);
const { data } = await response.json();

if (data.matchingAnalysis.matchPercentage >= 70) {
  console.log("Good match - move to interview");
} else if (data.matchingAnalysis.matchPercentage >= 40) {
  console.log("Fair match - consider for development role");
} else {
  console.log("Poor match - skill gaps significant");
}
```

### Example 2: Hiring Success Prediction

**Scenario**: Predict if a candidate will be successfully hired.

**Request**:
```
GET /api/ml/hiring-prediction?applicationId=app456
```

**Use Case**:
```javascript
// Prioritize applications by success probability
const predictions = await Promise.all(
  applicationIds.map(id => 
    fetch(`/api/ml/hiring-prediction?applicationId=${id}`)
      .then(r => r.json())
  )
);

const sorted = predictions.sort(
  (a, b) => b.data.successProbability - a.data.successProbability
);

// Prioritize top 20% for immediate action
const priority = sorted.slice(0, sorted.length * 0.2);
```

### Example 3: Get Recommendations

**Scenario**: Find best candidates for a specific proposal.

**Request**:
```
GET /api/ml/recommendations?proposalId=prop789&limit=5
```

**Response Processing**:
```javascript
const response = await fetch(
  `/api/ml/recommendations?proposalId=${proposalId}&limit=5`
);
const { data } = await response.json();

// Show candidates ranked by recommendation score
data.recommendations.forEach(rec => {
  console.log(`Candidate: ${rec.candidateId}`);
  console.log(`Match Score: ${rec.recommendationScore}%`);
  console.log(`Reason: ${rec.matchDetails.recommendationReason}`);
  console.log(`Gaps to fill: ${rec.gaps.join(", ")}`);
});
```

### Example 4: Batch Skill Matching

**Scenario**: Compare multiple student-proposal pairs at once.

**Request**:
```javascript
const response = await fetch('/api/ml/advanced-matching', {
  method: 'POST',
  headers: { 'Content-Type': 'application/json' },
  body: JSON.stringify({
    pairs: [
      { studentId: 'student1', proposalId: 'prop1' },
      { studentId: 'student2', proposalId: 'prop1' },
      { studentId: 'student3', proposalId: 'prop1' }
    ]
  })
});

const { data } = await response.json();
console.log(`Excellent matches: ${data.summary.excellentMatches}`);
console.log(`Good matches: ${data.summary.goodMatches}`);
```

## Best Practices

### 1. Skill Input Format
- **Keep skills consistent**: Use standardized skill names
- **Good**: "React", "Python", "SQL"
- **Avoid**: "react js", "PYTHON", "sql db"
- **Split related skills**: Use commas for multiple skills
- **Example**: "React, Redux, TypeScript" (not "React Redux")

### 2. Threshold Recommendations
- **Hire immediately**: Match ≥ 85%
- **Interview**: Match 70-85%
- **Skill development**: Match 50-70%
- **Not recommended**: Match < 50%

### 3. Using Confidence Levels
- **High confidence**: >6 months data, reliable predictions
- **Medium confidence**: 3-6 months data, use with caution
- **Low confidence**: <3 months data, validation needed

### 4. Forecasting Best Practices
- **Use 6-month forecasts** for immediate planning
- **12-24 month forecasts** are less reliable
- **Combine with business context** for decision-making
- **Monitor actual vs predicted** and adjust model

### 5. Handling Anomalies
- **High severity**: Immediate investigation required
- **Medium severity**: Schedule follow-up review
- **Low severity**: Monitor trend but no urgent action

### 6. Data Quality
- **Update skills regularly**: Stale data affects matching
- **Use consistent statuses**: Standardize pipeline stages
- **Complete application data**: All fields should be filled
- **Monitor for missing data**: Check database integrity

## Troubleshooting

### Issue 1: Low Skill Match Scores
**Symptom**: Matching scores are consistently below 50%

**Causes**:
- Skills are not normalized (different capitalization)
- Skills use domain-specific jargon
- Skill database has incomplete entries

**Solutions**:
1. Standardize skill names in your system
2. Create skill aliases mapping (e.g., "JS" → "JavaScript")
3. Review and update skill entries in database
4. Test with explicit skill arrays

### Issue 2: Forecasting Shows Flat Trend
**Symptom**: All forecasted months show similar numbers

**Causes**:
- Insufficient historical data (< 3 months)
- Very consistent application volume (no trend)
- Seasonal patterns not yet captured

**Solutions**:
1. Collect more data over longer period
2. Check data quality for gaps or errors
3. Use domain expertise to adjust predictions
4. Monitor actual vs predicted for validation

### Issue 3: High Variance in Predictions
**Symptom**: Confidence levels are very low

**Causes**:
- Insufficient data sample
- High volatility in hiring process
- External factors affecting recruitment

**Solutions**:
1. Collect more historical data
2. Investigate root causes of volatility
3. Use qualitative factors alongside predictions
4. Increase confidence threshold for decision-making

### Issue 4: Anomalies Triggering Too Often
**Symptom**: Too many false positive anomaly alerts

**Causes**:
- Thresholds set too sensitively
- Normal business fluctuations misidentified
- Seasonal patterns not accounted for

**Solutions**:
1. Review and adjust severity thresholds
2. Analyze historical anomalies for patterns
3. Consider business context when setting thresholds
4. Implement feedback loop to improve detection

### Issue 5: API Returning Error 500
**Symptom**: ML endpoints consistently fail

**Causes**:
- Database connection issues
- Missing required parameters
- Corrupted data in database
- Memory issues with large dataset

**Solutions**:
1. Check database connection health
2. Validate API request parameters
3. Verify data integrity in database
4. Check server logs for specific errors
